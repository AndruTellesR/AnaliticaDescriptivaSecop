# Notebook 04 — Logistic Regression

Algoritmo: Logistic Regression. Escalado: sí. n_iter = 40, CV=5, scoring='roc_auc'.

Observaciones: Línea base lineal. Necesita escalado.

## 1. Setup

```python
import sys, warnings
from pathlib import Path
warnings.filterwarnings('ignore')

RAIZ = Path('..').resolve()
sys.path.insert(0, str(RAIZ))

from src.preprocesamiento import preparar, RANDOM_STATE
from src.entrenamiento  import entrenar_y_evaluar

DATA = RAIZ / 'data'

from sklearn.linear_model import LogisticRegression
from scipy.stats import loguniform
```

## 2. Cargar y preparar datos

```python
datos = preparar(DATA, aplicar_escalado=True)
print(f'Train: {datos.n_train} | Test: {datos.n_test} | Features: {len(datos.cols_features)}')
print(f'scale_pos_weight atraso: {datos.scale_pos_weight["tuvo_atraso"]:.4f}')
```

## 3. Clasificador + grid

```python
clf = LogisticRegression(max_iter=2000, random_state=RANDOM_STATE, n_jobs=-1)

param_distributions = {
        'C': loguniform(0.01, 100),
        'penalty': ['l2'],
        'solver': ['lbfgs', 'liblinear'],
        'class_weight': ['balanced', None],
    }
```

## 4. RandomizedSearchCV (40 iters × CV 5)

```python
resultado = entrenar_y_evaluar(
    nombre='lr',
    estimador=clf,
    param_distributions=param_distributions,
    datos=datos,
    data_dir=DATA,
    n_iter=40,
    target='tuvo_atraso',
)
resultado
```
