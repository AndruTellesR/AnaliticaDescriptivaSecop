Algoritmo: K-Nearest Neighbors. Escalado: sí. n_iter = 25, CV=5, scoring='roc_auc'.

Observaciones: Distancia. Necesita escalado. Sensible a cardinalidad.

## 1. Setup

```python
import sys, warnings
from pathlib import Path
warnings.filterwarnings('ignore')

RAIZ = Path('..').resolve()
sys.path.insert(0, str(RAIZ))

from src.preprocesamiento import preparar, RANDOM_STATE
from src.entrenamiento  import entrenar_y_evaluar

DATA = RAIZ / 'data'

from sklearn.neighbors import KNeighborsClassifier
```

## 2. Cargar y preparar datos

```python
datos = preparar(DATA, aplicar_escalado=True)
print(f'Train: {datos.n_train} | Test: {datos.n_test} | Features: {len(datos.cols_features)}')
print(f'scale_pos_weight atraso: {datos.scale_pos_weight["tuvo_atraso"]:.4f}')
```

## 3. Clasificador + grid

```python
clf = KNeighborsClassifier(n_jobs=-1)

param_distributions = {
        'n_neighbors': [7, 11, 15, 21, 31, 51],
        'weights': ['uniform', 'distance'],
        'metric': ['euclidean', 'manhattan'],
    }
```

## 4. RandomizedSearchCV (25 iters × CV 5)

```python
resultado = entrenar_y_evaluar(
    nombre='knn',
    estimador=clf,
    param_distributions=param_distributions,
    datos=datos,
    data_dir=DATA,
    n_iter=25,
    target='tuvo_atraso',
)
resultado
```
