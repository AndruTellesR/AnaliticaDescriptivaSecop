# Notebook 07 — Random Forest

Algoritmo: Random Forest. Escalado: no. n_iter = 40, CV=5, scoring='roc_auc'.

Observaciones: Bagging. Sin escalado. Captura interacciones no-lineales.

## 1. Setup

```python
import sys, warnings
from pathlib import Path
warnings.filterwarnings('ignore')

RAIZ = Path('..').resolve()
sys.path.insert(0, str(RAIZ))

from src.preprocesamiento import preparar, RANDOM_STATE
from src.entrenamiento  import entrenar_y_evaluar

DATA = RAIZ / 'data'

from sklearn.ensemble import RandomForestClassifier
```

## 2. Cargar y preparar datos

```python
datos = preparar(DATA, aplicar_escalado=False)
print(f'Train: {datos.n_train} | Test: {datos.n_test} | Features: {len(datos.cols_features)}')
print(f'scale_pos_weight atraso: {datos.scale_pos_weight["tuvo_atraso"]:.4f}')
```

## 3. Clasificador + grid

```python
clf = RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1)

param_distributions = {
        'n_estimators': [200, 300, 500],
        'max_depth': [None, 10, 20, 30],
        'min_samples_split': [2, 5, 10],
        'min_samples_leaf': [1, 2, 4],
        'max_features': ['sqrt', 'log2'],
        'class_weight': ['balanced', 'balanced_subsample', None],
    }
```

## 4. RandomizedSearchCV (40 iters × CV 5)

```python
resultado = entrenar_y_evaluar(
    nombre='rf',
    estimador=clf,
    param_distributions=param_distributions,
    datos=datos,
    data_dir=DATA,
    n_iter=40,
    target='tuvo_atraso',
)
resultado
```
